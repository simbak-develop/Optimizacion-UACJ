# **CODIGO FUENTE**

```python
import pyomo.environ as pyo

# Crear el modelo
model = pyo.ConcreteModel()

# Variables de decisión
model.x1 = pyo.Var(within = pyo.NonNegativeIntegers) #Trio
model.x2 = pyo.Var(within = pyo.NonNegativeIntegers) #Maiz

# Función objetivo
model.obj =pyo.Objective(expr= 500 * model.x1 + 300 * model.x2,sense=pyo.maximize) #Ganancia


# Restricciones
model.con1 = pyo.Constraint(expr=model.x1 + model.x2 <=50)  #Limite total de Hectareas
model.con2 = pyo.Constraint(expr=model.x1 <=40) #Limite de Trigo
model.con3 = pyo.Constraint(expr=model.x2 >= 10) #Minimo de Maiz

model.cons_x1 = pyo.Constraint(expr=model.x1 >=0) #var x1
model.cons_x2 = pyo.Constraint(expr=model.x2 >=0) #var x2

# Resolver
solver = pyo.SolverFactory('glpk')
result = solver.solve(model)


# Mostrar resultados
print(f" Ganancia máxima : {model.obj():.2f}")
print(f"x1: {int(model.x1.value)} , x2: {int(model.x2.value)}")

```

## Problema de Optimización: Asignación de Cultivos

Un granjero tiene **50 hectáreas** de tierra disponibles y puede plantar **trigo** (`x₁`) o **maíz** (`x₂`).  
Cada hectárea de trigo genera **$500** de ganancia, y cada hectárea de maíz genera **$300**.

### Restricciones:
- No se pueden utilizar más de **50 hectáreas** en total.
- No se pueden dedicar más de **40 hectáreas al trigo**.
- Se deben dedicar **al menos 10 hectáreas al maíz**.

### Objetivo:
**Maximizar las ganancias.**


---

## **SOLUCION MATEMATICA**

## Formulación Matemática del Problema

Maximizar:  
$Z = 500x_1 + 300x_2$

Sujeto a:
$$
\left\{
\begin{array}{ll}
x_1 + x_2 & \leq 50 \quad \text{(Límite total de hectáreas)} \\
x_1 & \leq 40 \quad \text{(Límite de trigo)} \\
x_2 & \geq 10 \quad \text{(Mínimo de maíz)} \\
x_1, x_2 & \geq 0
\end{array}
\right.
$$
