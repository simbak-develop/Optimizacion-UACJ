## **CODIGO FUENTE**

```python
import pyomo.environ as pyo


# Crear el modelo
model = pyo.ConcreteModel()

#Variables  de decisión
model.x1 = pyo.Var(within=pyo.NonNegativeIntegers)
x1 = model.x1
model.x2 = pyo.Var(within=pyo.NonNegativeIntegers)
x2 = model.x2


#Función objetivo
model.Obj = pyo.Objective(expr= 40 * x1+ 30 * x2 , sense=pyo.maximize) #Ganancia

#Restricciones
model.Cons_Recurso_A = pyo.Constraint(expr= 2 * x1 + x2  <= 100) #Limite Recurso A
model.Cons_Recurso_B = pyo.Constraint(expr= x1 + 2 * x2 <= 80)  #Limite Recurso B

# model.cons_x1 = pyo.Constraint(expr= model.x1 >=0)
# model.cons_x2 = pyo.Constraint(expr= model.x2 >=0)
def no_nagative_values(model,varname):
    var = getattr(model,varname)
    setattr(model,f"cons_{varname}",pyo.Constraint(expr=var >=0))

no_nagative_values(model,x1.to_string())
no_nagative_values(model,x2.to_string())


#Resolver
solver = pyo.SolverFactory('glpk')
result = solver.solve(model)

# Mostrar resultado
print(f"Ganancia máxima: ${model.Obj():.2f}")
print(f" x1: {int(model.x1.value)} ,  x2: {int(model.x2.value)}")


```

## **PROBLEMA DE PRGRAMACION LINEAL**


Una fábrica produce dos tipos de mezclas, x1 y x2, con márgenes de ganancia de $40 y $30 por unidad, respectivamente. Cada unidad requiere cierta cantidad de recursos limitados para su producción:

* Recurso A: 2 unidades para x1 y 1 para x2, con un total de 100 disponibles.
* Recurso B: 1 unidad para x1 y 2 para x2, con un total de 80 disponibles.

Objetivo: Maximizar la ganancia total.

---

## **SOLUCION MATEMATICA**

Maximizar: $Z = 40x_1 + 30x_2$

Sujeto a:
$$
\left\{
\begin{array}{l} % Usamos 'll' para dos columnas, la primera para el inicio de la inecuación y la segunda para los términos
2x_1 + x_2 & \leq 100 \\
x_1 + 2x_2 & \leq 80 \\
x_1, x_2 & \geq 0
\end{array}
\right.
$$
