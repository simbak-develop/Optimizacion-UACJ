## **CODIGO FUENTE**

````python
from pyomo.environ import *

# Crear un modelo simple
model = ConcreteModel()
model.x = Var(bounds=(0,10),domain=Integers)
x = model.x
model.y = Var(bounds=(0,10),domain=Integers)
y = model.y

# Objective
model.Obj = Objective(expr = x + y, sense=minimize)

# Constraint
model.Cons = Constraint(expr=x + 2 * y >= 10)

# Solver con GLPK
solver = SolverFactory('glpk')
result = solver.solve(model)

# Resultados
if (result.solver.status == 'ok') and (result.solver.termination_condition == 'optimal'):
    print('Solución óptima encontrada')
    print(f'x = {x.value}')
    print(f'y = {y.value}')
    print(f'Objetivo = {model.Obj.expr()}') # Corrected line
else:
    print("Error al resolver el modelo")
    print(f'Solver status: {result.solver.status}, termination condition: {result.solver.termination_condition}')
````

## **DOCUMENTACION DE PRACTICA**


### **1. Importar la librería Pyomo**

from pyomo.environ import *

**Explicación**: Aquí estamos importando todas las funciones y clases de la librería pyomo.environ. Pyomo es una herramienta de modelado de optimización en Python, y pyomo.environ es un módulo que contiene las funciones básicas para definir y resolver problemas de optimización.



---



## **2. Crear un modelo simple**

model = ConcreteModel()

**Explicación**: Aquí estamos creando un objeto de tipo ConcreteModel. Un ConcreteModel es un modelo de optimización donde todas las componentes (variables, restricciones, etc.) se definen explícitamente. Es como un "contenedor" donde vamos a agregar todas las partes de nuestro problema de optimización.



---


## **3. Definir las variables del modelo**

model.x = Var(bounds=(0,10))

model.y = Var(bounds=(0,10))

**Explicación**: Aquí estamos definiendo dos variables, x e y, que son las variables de decisión en nuestro problema de optimización.


*  Var es una función que define una variable en el modelo.
*  bounds=(0,10) indica que las variables x e y pueden tomar valores entre 0 y 10. Es decir, x e y están acotadas entre 0 y 10.




---



## **4. Definir la función objetivo**

model.obj = Objective(expr= model.x + model.y, sense=minimize)


**Explicación**: Aquí estamos definiendo la función objetivo del problema de optimización.


*   Objective es una función que define la función objetivo.

*   expr= model.x + model.y es la expresión matemática que queremos optimizar. En este caso, queremos minimizar la suma de x e y.


*  sense=minimize indica que queremos minimizar la función objetivo. Si quisiéramos maximizar, usaríamos sense=maximize.




---


## **5. Definir una restricción**

model.cons = Constraint(expr=model.x + 2 * model.y >= 10)

**Explicación**: Aquí estamos definiendo una restricción en el modelo.

*   Constraint es una función que define una restricción en el modelo.
*   expr=model.x + 2 * model.y >= 10 es la expresión matemática que representa la restricción. En este caso, la suma de x y el doble de y debe ser mayor o igual a 10.



---

## **6. Seleccionar el solver y resolver el modelo**

solver = SolverFactory('glpk')

resultado = solver.solve(model)

**Explicación**: Aquí estamos seleccionando un solver (resolvedor) para resolver el modelo.



*  SolverFactory('glpk') crea un objeto que representa el solver GLPK (GNU Linear Programming Kit), que es un solver de optimización lineal.
*  solver.solve(model) resuelve el modelo utilizando el solver seleccionado. El resultado de la solución se almacena en la variable resultado.



---

## **7. Mostrar los resultados**

```python
if resultado.solver.status == SolverStatus.ok:

    print(f"x: {model.x.value}, y: {model.y.value}")
else:

*  resultado.solver.status == SolverStatus.ok verifica si el solver encontró una solución exitosa.
*   Si la solución fue exitosa, se imprimen los valores de x e y que minimizan la función objetivo.
